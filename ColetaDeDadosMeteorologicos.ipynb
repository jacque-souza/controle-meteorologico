```python
# https://docs.google.com/spreadsheets/d/1ZQjYKCPDoORHRCziTOCa3XhdKgGUe_zgQU5wRg9d-4c/edit?resourcekey=&gid=493950749#gid=493950749

import pandas as pd
import requests
import time

endpoint_planilha = "https://docs.google.com/spreadsheets/d/1ZQjYKCPDoORHRCziTOCa3XhdKgGUe_zgQU5wRg9d-4c/export?format=csv"
endpoint_api = "https://api.open-meteo.com/v1/forecast"
```

```python
print('---- Dados Coletados ----')
dados_planilha = pd.read_csv(endpoint_planilha)

display(dados_planilha)
```

```python
# iterar cada resposta do formulario
for indice, linha in dados_planilha.iterrows():
  nome_local = linha['Nome do local']
  lat = linha['Latitude']
  lon = linha['Longitude']

  parametros ={
      'latitude':lat,
      'longitude':lon,
      'current': 'temperature_2m'
  }

  resposta = requests.get(endpoint_api, params=parametros).json()

  display(resposta)
  print('\n\n')
```
